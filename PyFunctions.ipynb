```python
# *args(any no of arguments) and
# **kwargs(any no of KEYWORD args)
# *args and **kwargs allow functions to accept a unknown number of arguments.
def anyNoOfArgs(*args):
  for x in args:
    print(x)

anyNoOfArgs(1,2,3)
```

```python
#**kwargs
def anyNoOfKewwordArgs(**args):
  for x in args:
    print(x, args[x])

anyNoOfKewwordArgs(a=1,b=2,c=3)
```

```python
# Iterative approach for summation
def sum_iterative(n):
  total = 0
  for i in range(1, n + 1):
    total += i
  return total

number_to_sum = 5
print(f"Summation using iteration for {number_to_sum}: {sum_iterative(number_to_sum)}")
```

You can also use a mathematical formula for the sum of the first 'n' natural numbers, which is `n * (n + 1) / 2`. This is often the most efficient method for large numbers.

```python
# Mathematical formula for summation
def sum_formula(n):
  return n * (n + 1) // 2

number_to_sum = 10
print(f"Summation using formula for {number_to_sum}: {sum_formula(number_to_sum)}")
```

#Generators

Generators are functions that can pause and resume their execution.

When a generator function is called, it returns a generator object, which is an iterator.

The code inside the function is not executed yet, it is only compiled. The function only executes when you iterate over the generator.

The yield keyword is what makes a function a generator.

When yield is encountered, the function's state is saved, and the value is returned. The next time the generator is called, it continues from where it left off.

```python
def generator():
  yield 1
  yield 2
  yield 3

for value in generator():
  print(value)
#print(generator())
#print(list(generator())
```

```python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1

for num in count_up_to(5):
  print(num)
```

```python
# Iterator
#An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.
# is an object which implements the iterator protocol, which consist of the methods __iter__() and __next__()
'''
Lists, tuples, dictionaries - iterable
All these objects have a iter() - iterator
'''
```

```python
class MyNumbers:
  def __iter__(self):
    self.a = 1
    return self

  def __next__(self):
    if self.a <= 20:
      x = self.a
      self.a += 1
      return x
    else:
      raise StopIteration

myclass = MyNumbers()
myiter = iter(myclass)

for x in myiter:
  print(x)
```
