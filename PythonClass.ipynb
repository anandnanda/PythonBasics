#Inheritance

```python
class ParentClass:
    def __init__(self,name,age=20):
     self.name=name
     self.age=age

    def __str__(self):
      return f'{self.name}, {self.age}'

class ChildClass1(ParentClass):
  pass

c1=ChildClass1('AAA',30)
print(c1)
```

```python

class ChildClass3(ParentClass):
  def __init__(self,name,age,graduationYr):
    ParentClass.__init__(self,name,age)
    self.graduationYr=graduationYr

c2=ChildClass3('BBB',20,2020)
print(c2)
print(c2.graduationYr)
```

```python
#super() function, you do not have to use the name of the parent element, it will automatically inherit the methods and properties from its parent.
class ChildClass2(ParentClass):
  def __init__(self,name,age,graduationYr):
    super().__init__(name,age)
    self.graduationYr=graduationYr

c2=ChildClass2('BBB',20,2020)
print(c2)
print(c2.graduationYr)
```

```python
#Class polymorphism
class Vehicle:
  def __init__(self,name,color):
    self.name=name
    self.color=color
  def movesHow(self):
    print('Drives')

class Car(Vehicle):
  pass

class Boat(Vehicle):
  def __init__(self,name, color):
    Vehicle.__init__(self,name,color)
  def movesHow(self):
    print('Sails')

class Plane(Vehicle):
  def __init__(self,name,color):
    super().__init__(name,color)
  def movesHow(self):
    print('Flies')

c=Car('Car','Red')
b=Boat('Boat','Blue')
p=Plane('Plane','White')
print(c.name,c.color)
print(c.movesHow())
print(b.name,b.color)
print(b.movesHow())
print(p.name,p.color)
print(p.movesHow())
```

#Python Encapsulation
Encapsulation is about protecting data inside a class.

It means keeping data (properties) and methods together in a class, while controlling how the data can be accessed from outside the class.

Data Protection: Prevents accidental modification of data

Validation: You can validate data before setting it

Flexibility: Internal implementation can change without affecting external code

Control: You have full control over how data is accessed and modified

```python
# make properties private by using a double underscore __ prefix:
class Encapsulate:
  def __init__(self, name, age):
    self.name=name
    self.__age=age

  def printProtectedVar(self):
    print(self.__age)

  def updateProtectedVar(self,addAge):
    self.__age +=addAge
    print(self.__age)

e1=Encapsulate('AAA',20)
e1.printProtectedVar()
e1.updateProtectedVar(10)

#err
print(e1.__age)
```

```python
#this will be considered as diff variable
e2=Encapsulate('BBB',3)
e2.__age=40
e2.__agee=50
print(e2.__age, e2.__agee)
e2.printProtectedVar()
```
