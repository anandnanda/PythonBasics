```python
# With the subplot() function you can draw multiple plots in one figure
#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2,1) # row, col, plot no - 1
plt.plot(x,y)

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2,2) ## row, col, plot no - 2
plt.plot(x,y)

plt.show()
```

```python
plt.suptitle("Super title")

plt.plot(2,1,1)
plt.plot(xpoints,ypoints)
plt.title('plot1')
plt.show()

plt.plot(2,1,2)
plt.plot(x,y)
plt.title('plot2')

plt.show()
```

```python
# alpha and random color
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = 10 * np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')

plt.colorbar()

plt.show()
```

```python
#bar graph
plt.bar(x,y, color='hotpink')
plt.show()
```

```python
plt.barh(x,y, height=5)
plt.show()
```

#Histogram
A histogram is a graph showing frequency distributions.

It is a graph showing the number of observations within each given interval.

Example: Say you ask for the height of 250 people, you might end up with a histogram like this:

```python
h=np.random.normal(170,10,250)  #locaiton: 170 is say mean, scale:10 - plot range in y axis, size:250
print(h)
plt.hist(h)
plt.show()
```
